```python
import numpy as np
```

```python
viewMaskFromPaper = 1;

#total mode groups
maxMG = 20
modeCount = sum(1:maxMG)

lambda0 = 1565e-9 #centre wavelength

downsample = 4 #to downsample the sim N times from full resolution

pixelSize = downsample*8e-6 #phase mask pixel pitch

planeSpacing = 25.14e-3;

#How far is the input SMF array from the first plane?
arrayDistToFirstPlane = 20e-3;

#total number of planes
planeCount = 7

#Pixel counts of the masks and simulation in x and y dimensions
if (viewMaskFromPaper):
    Nx = 512./downsample
    Ny = 384./downsample
else:
    Nx = 256./downsample
    Ny = 256./downsample

#number of total passes to run for
if (viewMaskFromPaper):
    iterationCount = 1
else:
    iterationCount = 100

#Display graphs of fields/masks every N passes
graphIterations = 10

#Mode-field diameter (MFD) of input Gaussian beams
MFDin = 60e-6

#Mode-field diameter (MFD) of the output Hermite-Gaussian beams
MFDout = 400e-6
```

```python
#SIMULATION CONSTRAINTS
#Angle-space filter. Only propagate angles less than kSpaceFilter*(the
#maximum angle supported by the simulation. Which would in turn be given by
#the pixel size)
#e.g. 0.5 means only half the possible angle-space is used. Limiting angle
#space encourages solutions which are high-bandwidth, have low scatter,
#smooth and don't accidently wrap-around the edge of the simulation.

kSpaceFilter = 1000

maskOffset = np.sqrt(1e-3/(Nx*Ny*modeCount))
symmetricMasks = 1

#Creating the mask
X = ((np.arange(1, Ny + 1) - (Ny / 2 + 0.5))) * pixelSize
Y = ((np.arange(1, Nx + 1) - (Nx / 2 + 0.5))) * pixelSize
X, Y = np.meshgrid(X, Y)

```

```python
#Create the HG-modes
print('Generating output basis...');
#Convert to polar-coordinates and rotate by 45degrees
[TH, R] = np.cart2pol(X,Y)
[X0,Y0] = np.pol2cart(TH-np.pi/4,R)
#Generate a set of HG modes up to mode-group maxMG
[MODES,M,N, MODES_TOTAL] = np.generateBasisHG(maxMG,X0,Y0,MFDout);
```
