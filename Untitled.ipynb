```python
#Just chekcing communication with the DMD
import numpy as np
from ALP4 import *
from packages.dmd_holograms import parallel_lee
import time


# Connect to the ALP controller
alp_dir = "packages/ALP-4.3"
DMD = ALP4(version = '4.3', libDir=alp_dir)

# Initialize
DMD.Initialize()

print("DMD initialized")
print("Resolution:", DMD.nSizeX, "x", DMD.nSizeY)
```

```python
#Displaying all black pixels
bitDepth = 1

alp_dir = "packages/ALP-4.3"
DMD = ALP4(version = '4.3', libDir=alp_dir)

# Initialize
DMD.Initialize()

imgBlack = np.zeros(
    (DMD.nSizeY, DMD.nSizeX),
    dtype=np.uint8
)



print("DMD initialized")
print("Resolution:", DMD.nSizeX, "x", DMD.nSizeY)

DMD.SeqAlloc(nbImg=1, bitDepth=bitDepth)
DMD.ProjControl(ALP_PROJ_INVERSION, 1)
DMD.SeqPut(imgData=imgBlack.ravel())

DMD.SetTiming(illuminationTime=20000)  # 20 ms = 50 Hz
DMD.Run()
input("Black pattern is running. Press Enter to stop...")

DMD.Halt()
DMD.FreeSeq()
DMD.Free()
```

```python
#Displaying all white pixels

alp_dir = "packages/ALP-4.3"
DMD = ALP4(version = '4.3', libDir=alp_dir)

# Initialize
DMD.Initialize()

print("DMD initialized")
print("Resolution:", DMD.nSizeX, "x", DMD.nSizeY)

imgWhite = np.ones(
    (DMD.nSizeY, DMD.nSizeX),
    dtype=np.uint8
) * 255

#if I do not use ALP_PROJ_INVERSION then the projection is white when I display all black and vice-versa.

DMD.SeqAlloc(nbImg=1, bitDepth=1)
DMD.ProjControl(ALP_PROJ_INVERSION, 1) #inverts all the pixels - white to black and black to white
DMD.SeqPut(imgData=imgWhite.ravel())

DMD.SetTiming(illuminationTime=20000)
DMD.Run()

input("White pattern is running. Press Enter to stop...")

DMD.Halt()
DMD.FreeSeq()
DMD.Free()
```

Sending in the hologram now...

```python

alp_dir = "packages/ALP-4.3"
DMD = ALP4(version='4.3', libDir=alp_dir)

DMD.Initialize()

print("DMD initialized")
print("Resolution:", DMD.nSizeX, "x", DMD.nSizeY)

# --------------------------------------------------
# Create target complex field
# --------------------------------------------------

ny = DMD.nSizeY
nx = DMD.nSizeX

x = np.arange(nx) - nx/2
y = np.arange(ny) - ny/2
X, Y = np.meshgrid(x, y)

# Example Gaussian amplitude
sigma = 300
amplitude = np.exp(-(X**2 + Y**2) / (2*sigma**2))

# Example phase
phase = np.zeros_like(amplitude)

# Complex field
field = amplitude * np.exp(1j * phase)

# --------------------------------------------------
# Generate binary Lee hologram
# --------------------------------------------------

holo = parallel_lee(
    field,
    nuvec=(1/4, 0)
)

# Convert bool -> uint8
holo = holo.astype(np.uint8) * 255

print("Hologram shape:", holo.shape)
print("Hologram values:", np.unique(holo))

# --------------------------------------------------
# Send hologram to DMD
# --------------------------------------------------

DMD.SeqAlloc(
    nbImg=1,
    bitDepth=1
)

DMD.ProjControl(
    ALP_PROJ_INVERSION,
    1
)

DMD.SeqPut(
    imgData=holo.ravel()
)

DMD.SetTiming(
    illuminationTime=20000
)

DMD.Run()

input("Hologram is running. Press Enter to stop...")

DMD.Halt()
DMD.FreeSeq()
DMD.Free()
```

```python
import numpy as np
from ALP4 import ALP4

alp_dir = "packages/ALP-4.3"
DMD = ALP4(version='4.3', libDir=alp_dir)

DMD.Initialize()

print("DMD initialized")
print("Resolution:", DMD.nSizeX, "x", DMD.nSizeY)

# Circular aperture - illumination mask : a hologram can be imposed on this

nx = DMD.nSizeX
ny = DMD.nSizeY

x = np.arange(nx) - nx / 2
y = np.arange(ny) - ny / 2
X, Y = np.meshgrid(x, y)
radius = 300  # DMD pixels
circle = (X**2 + Y**2 <= radius**2)
field = circle.astype(np.complex128)

holo = parallel_lee(field, nuvec=(1/4, 0))

# holo = holo & circle

# DMD needs uint8 for the image
img = holo.astype(np.uint8) * 255

print("Image shape:", img.shape)
print("Pixel values:", np.unique(img))

# --------------------------------------------------
# Send to DMD
# --------------------------------------------------

DMD.SeqAlloc(nbImg=1, bitDepth=1)

DMD.ProjControl(ALP_PROJ_INVERSION, 1)

DMD.SeqPut(imgData=img.ravel())

DMD.SetTiming(illuminationTime=20000)

DMD.Run()

input("Circular beam pattern is running. Press Enter to stop...")

DMD.Halt()
DMD.FreeSeq()
DMD.Free()
```

```python
import numpy as np
from ALP4 import ALP4

alp_dir = "packages/ALP-4.3"
DMD = ALP4(version='4.3', libDir=alp_dir)

DMD.Initialize()

print("DMD initialized")
print("Resolution:", DMD.nSizeX, "x", DMD.nSizeY)

#Square
height, width = 1600, 2560
side = 600
half_side = side//2
cy, cx = height // 2, width // 2

square = np.zeros((height,width), dtype=np.float64)
square[cy - half_side : cy + half_side, cx - half_side : cx + half_side] = 255
field = square

holo = parallel_lee(field, nuvec=(1/4, 0))

# holo = holo & circle

# DMD needs uint8 for the image
img = holo.astype(np.uint8) * 255

print("Image shape:", img.shape)
print("Pixel values:", np.unique(img))

# --------------------------------------------------
# Send to DMD
# --------------------------------------------------

DMD.SeqAlloc(nbImg=1, bitDepth=1)

DMD.ProjControl(ALP_PROJ_INVERSION, 1)

DMD.SeqPut(imgData=img.ravel())

DMD.SetTiming(illuminationTime=20000)

DMD.Run()

input("square pattern pattern is running. Press Enter to stop...")

DMD.Halt()
DMD.FreeSeq()
DMD.Free()
```
